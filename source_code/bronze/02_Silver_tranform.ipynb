```python
df = spark.read.table("workspace.default.bmtc_routes")
```

```python
display(df)
```

```python
df_clean = df.dropna(subset = ["route_id"])
```

```python
df_clean = df_clean.toDF(*[c.lower().replace(" ","_")
                           for c in df_clean.columns])
```

```python
from pyspark.sql.functions import current_timestamp
df_clean = df_clean.withColumn("ingestion_at",current_timestamp())
```

```python
df_clean = df_clean.dropDuplicates()
```

```python
df_clean.write.format("delta").mode("overwrite").saveAsTable("workspace.default.silver_bmtc_routes")
```

quality check

```python
bronze_count = spark.sql("SELECT COUNT(*) FROM workspace.default.bmtc_routes").collect()[0][0]
silver_count = df_clean.count()
print(f"Bronze_rows : {bronze_count}")
print(f"silver_count : {silver_count}")
print(f"Rows_difference : {bronze_count - silver_count}")
```
